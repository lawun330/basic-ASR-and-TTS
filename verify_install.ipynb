# Verify Installation

## Step 1: Install

1. Install PyTorch that supports Nvidia GTX 1060:

```bash
pip install torch torchaudio --index-url https://download.pytorch.org/whl/cu126
```

2. Install other packages:

```bash
pip install -r requirements.txt
```

3. Install vocoders:

```python
import torch

# install HiFi-GAN
torch.hub.load('NVIDIA/DeepLearningExamples:torchhub', 'nvidia_hifigan', pretrained=True, trust_repo=True)

# install 
torch.hub.load('NVIDIA/DeepLearningExamples:torchhub', 'nvidia_waveglow', pretrained=True, trust_repo=True)
```

4. Install text-rendering for mel images:

```bash
sudo apt update

# richer title rendering
sudo apt install python3-gi python3-gi-cairo gir1.2-pango-1.0

# Myanmar fonts
sudo apt install fonts-sil-padauk fonts-noto-core
```
